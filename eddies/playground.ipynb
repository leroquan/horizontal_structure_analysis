```python
import numpy as np
import pandas as pd
import os
import matplotlib.pyplot as plt
```

```python
zug_path = rf"/storage/alplakes_test/zug_100m_2025"
zug_test_path = rf"/storage/alplakes_test/zug_test_100m_2025"
```

```python
df_ke_zug = pd.read_csv(os.path.join(zug_path, "outputs_swirl", "ke_eddy", "ke_lake.csv"))
df_ke_zug_test = pd.read_csv(os.path.join(zug_test_path, "outputs_swirl", "ke_eddy", "ke_lake.csv"))
```

```python
df_ke_zug['date'] = pd.to_datetime(df_ke_zug['date'])
df_ke_zug_test['date'] = pd.to_datetime(df_ke_zug_test['date'])
df_ke_zug = df_ke_zug.set_index('date')
df_ke_zug_test = df_ke_zug_test.set_index('date')
```

```python
df_ke_zug['kinetic_energy_[MJ]'].plot(label='Base')
df_ke_zug_test['kinetic_energy_[MJ]'].plot(label='Test')
plt.xlim('2025-06-01', '2025-08-31')
plt.ylim(0,1700)
plt.legend()
```

```python
df_ke_eddy_zug = pd.read_csv(os.path.join(zug_path, "outputs_swirl", "ke_eddy", "ke_eddies.csv"))
df_ke_eddy_zug_test = pd.read_csv(os.path.join(zug_test_path, "outputs_swirl", "ke_eddy", "ke_eddies.csv"))
```

```python
df_ke_eddy_zug['date'] = pd.to_datetime(df_ke_eddy_zug['date'])
df_ke_eddy_zug_test['date'] = pd.to_datetime(df_ke_eddy_zug_test['date'])
df_ke_eddy_zug = df_ke_eddy_zug.set_index('date')
df_ke_eddy_zug_test = df_ke_eddy_zug_test.set_index('date')
```

```python
df_ke_eddy_zug['kinetic_energy_eddy_[MJ]'].plot(label='Base')
df_ke_eddy_zug_test['kinetic_energy_eddy_[MJ]'].plot(label='Test')
plt.xlim('2025-06-01', '2025-08-31')
plt.ylim(0,200)
plt.legend()
```

```python
(100 * df_ke_eddy_zug['kinetic_energy_eddy_[MJ]'] / df_ke_zug['kinetic_energy_[MJ]']).plot(label='Base')
(100 * df_ke_eddy_zug_test['kinetic_energy_eddy_[MJ]'] / df_ke_zug_test['kinetic_energy_[MJ]']).plot(label='Test')
plt.xlim('2025-07-01', '2025-08-01')
plt.legend()
```
